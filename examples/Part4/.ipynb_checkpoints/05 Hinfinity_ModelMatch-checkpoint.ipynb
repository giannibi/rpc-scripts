### The model matching problem
Given the plant $P$, we want to design a controller $K$ so that the closed-loop transfer function $T$ from the reference $r$ to the output $\tilde y$ (complementary sensitivity) closely matches a given closed loop $T_0$. This way, if the reference transfer function $T_0$ is designed so as to satisfy some performance requirements (for example, on the rise time $t_s$ and overshoot $\hat s$), achieving $T\approx T_0$ guarantees that the specifications are satisfied. This can be formaulated in an $H_\infty$ context as the minimization of the model matching error $||T-T_0||_\infty$. To formulate this as the standard problem, we need to define a suitable performance input $w$ and performance output $z_1$ so that $T-T_0$ is precisely the transfer function from $w$ to $z_1$. Choosing $w$ as the reference signal $r$, and being $T$ the transfer function from $r$ to $\tilde y$, we can define $z$ as the difference between $\tilde y$ and the output $\tilde y_0$ of $T_0$ with input $r$. For added flexibility, it is convenient to consider $||W_o(T-T_0)||_\infty$ as the performance metric, where $W_o$ is some stable filter. This way, a closer matching between $T$ and $T_0$ can be enforced at lower frequencies (within the bandwidth of $T$), where matching is more important in order to achieve the same $t_s$ and $\hat s$ as $T_0$, while leaving room for mismatch at higher frequencies to accommodate additional requirements, for example on the command sensitivity. Therefore, input $w$ is chosen as the reference signal $r$ and the performance output as $z_1=W_o(\tilde y-\tilde y_0)$. Computing as usual the transfer function from $[w\ u]$ to $[z_1\ y]$ yields the LFT for this problem:
$$
\begin{bmatrix}
z_1 \\
y
\end{bmatrix}
=
G
\begin{bmatrix}
w \\
u
\end{bmatrix}
\quad
G =
\begin{bmatrix}
- W_oT_0 & W_oP \\
I & -P
\end{bmatrix}
$$

We can also include as an additional performance requirement, the standard bound on the reference-to-command transfer function $KS$, we add a performance weight $W_u$ on the command signal $u$, obtaining a $z_2=W_u u$ performance signal and we take the latter into account in the performance outputs so that $
z =
\begin{bmatrix}
z_1 \\
z_2
\end{bmatrix}
$, with the corresponding $G$ matrix
$$
G =
\begin{bmatrix}
- W_oT_0 & W_oP \\
0 & W_u \\
I & -P
\end{bmatrix}
$$


### Point B: The plant
---
Considering the elastic robot joint model we have a coupling between motor and payload, so we can write down the system of equations for each component:


\begin{cases}
J \ddot{\theta} + B (\dot{\theta} - \dot{q}) + K (\theta - q) = u   \quad Motor \\
M \ddot{q} + B (\dot{q} - \dot{\theta}) + K (q - \theta) = 0 \quad Payload
\end{cases}

and obtaing the following transfer function between the joint position and the commanded torque:
$$
P(s)=\frac{Q(s)}{U(s)}=\frac{B s + K}{s^2 \left( J M s^2 + (B J + B M) s + (J K + K M) \right)}
$$


```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct
import slycot as sly
import cvxpy as cp
from itertools import product
import rpc_diism.robust as rb
```

```python
# Plant Parameters

J = 0.01
K = 7.3 
B = 0.04 
M = 0.009
```

```python
#Frequency range
omega = np.logspace(-2, 4, 1000)
```

```python
# Plant P
P = ct.tf([B, K],[J*M, B*M+B*J, K*M+K*J,0,0])
display("Plant tf P", P)
```

```python
# Plant is corrected: poles shifted to avoid numerical problems since the original poles on the imaginary axis
s=ct.tf('s')
# Natural frequency of the dominant poles of the plant
wp = 27 
epsc = 1e-6

sc =s+wp*epsc
Pc = (K+B*sc)/(sc**4*(J*M)+sc**3*((M*B+B*J))+sc**2*(M*K+K*J))

display("Plant tf P corrected", Pc)
```

```python
#Comparing the Bode plot we can see the almost perfect overlap between the original plant with that of the shifted one
m1, p1, _ = P.frequency_response(omega)
m2, p2, _ = Pc.frequency_response(omega)
plt.figure(figsize=(16, 2))
plt.semilogx(omega, 20*np.log10(m1), 'm', label=r'$\P$')
plt.semilogx(omega, 20*np.log10(m2), 'c--', label=r'$\Pc)$')
```

### Performance specifications
---
The ideal matching function is a second order system for which the natural frequency $\omega_n$ and the damping $\zeta$ must be chosen in such a way as to respect the overshoot $\hat{s}$ and rise time $t_s$, a damping of $\zeta$=0.6 and $\omega_n$ = 3.7 is chosen.

$$
T_0(s)=\frac{1}{1+\frac{2\zeta s}{\omega_n}+\frac{s^2}{\omega_n^2}}, \quad \omega_n=3,7, \quad \zeta=0.6
$$

```python
# Matching function T_0 of 2° order
damping=0.6 
wn = 3.7 
T_0_num=[1]
T_0_den=[1/wn**2,(2*damping)/wn,1]
T_0= ct.tf(T_0_num,T_0_den)
#display("T_0 matching function", T_0)
```

```python
t, step_ = ct.step_response(T_0,np.linspace(0,5,1000))
plt.figure(figsize=(16, 3))
plt.plot(t, step_, label="Step_response", linewidth=1.2,color='m')
plt.axhline(y=1, color='b', linestyle='--', linewidth=1, label="")   
plt.axhline(y=1.1, color='g', linestyle='--', linewidth=1, label=r"$\hat{S}^{max}$") 

plt.xlabel("Time")
plt.title(r'Step response of matching function $T_0$')

plt.legend()
plt.show()
```

Weight on $KS$. The parameters for this weight are chosen to ensure that the controller can effectively match the desired dynamics, modeled by $T_0$, within its bandwidth, so $wbc$ is taken beyond $w_n$. A multiple pole-zero version is used.

```python
# Weight on KS (command)
M_u = 1000
eps = 0.001
wbc = 30*wn
Wu = ct.ss(rb.weightU(wbc, M_u,eps,5))
```

Model matching weight $W_o$. Since we want to match the transient, it is necessary that the difference is minimum in the frequency range corresponding to the bandwith. The weight that is suitable to achieve this goal is of the same shape as the weight $W_e$ that usually weighs the sensitivity function $S$. The choice of parameters ensures that at low frequencies the weight is very high forcing the closed loop to closely follow the desired behavior. At high frequencies, the weight allows greater flexibility where matching is less critical.

```python
# Weight on matching
Mo = 10
wbo = 10*wn
epso = 1e-3
Wo = ct.ss(rb.weightS(wbo,Mo,epso,1))
```

```python
# Singular values of weights
Wu_inv_sv = rb.sigma(rb.invss(Wu), omega)
Wo_inv_sv = rb.sigma(rb.invss(Wo), omega)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 3))

axes[0].semilogx(omega, 20*np.log10(Wu_inv_sv), 'r--', label=r'$\sigma_1({W_u}^{-1})$')
axes[0].set_ylim([-100,70])
axes[0].set_xlim([1e-2,1e3])
axes[0].set_ylabel('magnitude [dB]')
axes[0].set_xlabel('freq [rad/s]')
axes[0].legend()
axes[0].set_title(r'Singular values of $W_u^{-1}$')

axes[1].semilogx(omega, 20*np.log10(Wo_inv_sv), 'r--', label=r'$\sigma_1({W_o}^{-1})$')
axes[1].set_ylabel('magnitude [dB]')
axes[1].set_xlabel('freq [rad/s]')
axes[1].legend()
axes[1].set_title(r'Singular values of $W_o^{-1}$')
```

Computation of the LFT for the H-infinity problem

```python
#Matching function block
T_0.input_labels = ['r']
T_0.output_labels = ['yo']
#Correct plant block
fbk = ct.summing_junction(inputs=['r','-yp'], outputs=['y'])
#Matching weight block
Wu.input_labels = ['u']
Wu.output_labels = ['ubar']
#Command input block
Pc.input_labels = ['u']
Pc.output_labels = ['yp']
#Difference matching
sum = ct.summing_junction(inputs=['yp','-yo'], outputs=['ye'])
# Weight matching
Wo.input_labels = ['ye']
Wo.output_labels = ['ybar']
#G block
G = ct.minreal(ct.interconnect([T_0,Pc, Wo, Wu, fbk,sum],inputs=['r','u'],outputs=['ybar','ubar','y']))
```

```python
K_c, Tzw, gamma, rcond = rb.hinfsyn(G, nmeas=1, ncon=1)
display("gamma:", gamma)
```

### Evaluation of performance

```python
# Complementary function of nominal plant and matching
fig, axes = plt.subplots(1, 4, figsize=(18, 3))
#1) KS vs gamma*sigma1(Wu^-1)
# Ks on input command
S = ct.feedback(1, Pc * K_c)
KS_sv = rb.sigma(K_c * S, omega)

Wu = ct.ss(rb.weightU(wbc, M_u, eps,5))
Wu_inv = rb.invss(Wu)
Wuinv_sv = rb.sigma(Wu_inv, omega) * gamma
axes[0].semilogx(omega, 20 * np.log10(KS_sv), 'g', label=r'$\sigma(KS)$')
axes[0].semilogx(omega, 20 * np.log10(Wuinv_sv), 'm:', label=r'$\gamma \sigma({Wu}^{-1})$')
axes[0].fill_between(omega, -6, -120, where=(omega > 100), color='c', alpha=0.6)
axes[0].set_ylim([-100,110])
axes[0].set_xlim([1e-2,1e3])
axes[0].set_ylabel('Magnitude [dB]')
axes[0].set_xlabel('Frequency [rad/s]')
axes[0].legend()
axes[0].set_title(r'Singular values of KS vs.  $\gamma \sigma({Wu}^{-1})$')

#2) T-T0 vs gamma*sigma1(Wo^-1)
T = ct.feedback(Pc * K_c)
matching = T-T_0
match_sv = rb.sigma(matching, omega)

Wo = ct.ss(rb.weightS(wbo, Mo, epso,1))
Wo_inv = rb.invss(Wo)
Woinv_sv = rb.sigma(Wo_inv, omega) * gamma
axes[1].semilogx(omega, 20 * np.log10(match_sv), 'g', label=r'$\sigma(T-T_0)$')
axes[1].semilogx(omega, 20 * np.log10(Woinv_sv), 'm:', label=r'$\gamma \sigma({Wo}^{-1})$')
axes[1].set_ylabel('Magnitude [dB]')
axes[1].set_xlabel('Frequency [rad/s]')
axes[1].legend()
axes[1].set_title(r'Singular values of (T-$T_0$) vs.  $\gamma \sigma({Wo}^{-1})$')

#3) Frequency response comparison between T and T0
m1, p1, _ = T.frequency_response(omega)
m2, p2, _ = T_0.frequency_response(omega)

axes[2].semilogx(omega, 20 * np.log10(m1), 'm', label=r'$T$')
axes[2].semilogx(omega, 20 * np.log10(m2), 'c', label=r'$T_0$')
axes[2].set_ylabel('Magnitude [dB]')
axes[2].set_xlabel('Frequency [rad/s]')
axes[2].legend()
axes[2].set_title(r'Frequency Response of T and $T_0$')

t, step_ = ct.step_response(T, np.linspace(0, 3, 1000))
axes[3].plot(t, step_, label="Step_response", linewidth=1.2, color='m')
axes[3].axhline(y=1, color='b', linestyle='--', linewidth=1, label="")   
axes[3].axhline(y=1.1, color='g', linestyle='--', linewidth=1, label=r"$\hat{S}^{max}$") 
axes[3].set_xlabel("Time")
axes[3].set_title(r'Step response of T')
axes[3].legend()

plt.tight_layout()
plt.show()
```
